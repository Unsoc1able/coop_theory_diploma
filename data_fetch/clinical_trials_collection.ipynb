# Руководство по сбору данных с ClinicalTrials.gov

Этот ноутбук описывает процесс обновления агрегированных данных по клиническим исследованиям для дашборда кооперативной модели R&D.

## Подготовка окружения

1. Установите зависимости проекта (см. `requirements.txt`).
2. Создайте и активируйте виртуальное окружение Python.
3. Убедитесь, что у вас есть доступ в интернет: загрузка данных выполняется через публичный API ClinicalTrials.gov.

```python
%%bash
python -m venv .venv
source .venv/bin/activate
pip install -r requirements.txt
```

## Обновление набора данных

Скрипт `data_fetch/build_clinical_trials_dataset.py` формирует файл `data/clinical_trials_by_sponsor.json`.\nЭтот файл автоматически подхватывается приложением. По умолчанию будут использованы спонсоры из списка `DEFAULT_SPONSORS`, но его можно переопределить через параметры командной строки.

```python
%%bash
python data_fetch/build_clinical_trials_dataset.py \
    --max-studies 500 \
    --batch-size 200 \
    --out data/clinical_trials_by_sponsor.json
```

### Кастомный набор компаний

Если необходимо ограничить сбор конкретными компаниями, добавьте несколько опций `--sponsor`. Скрипт допускает совместное использование международных и российских компаний.

```python
%%bash
python data_fetch/build_clinical_trials_dataset.py \
    --sponsor Pfizer \
    --sponsor BIOCAD \
    --sponsor "Johnson & Johnson" \
    --max-studies 300 \
    --out data/clinical_trials_by_sponsor.json
```

## Проверка и анализ

После успешного запуска убедитесь, что файл `data/clinical_trials_by_sponsor.json` содержит блок `companies`. Его можно быстро просмотреть из ноутбука.

```python
from pathlib import Path
import json

dataset_path = Path("data/clinical_trials_by_sponsor.json")
if dataset_path.exists():
    data = json.loads(dataset_path.read_text(encoding="utf-8"))
    companies = data.get("companies") or []
    companies[:5]
```
